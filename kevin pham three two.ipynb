```python
import numpy as np
import matplotlib.pyplot as plt

s = 0.1; N = 100; d = 0.01

x = np.zeros(N)
y = np.zeros(N)

x[0] = 1
y[0] = 1

k = 1
for i in range(1,N):
    xinc = -s * (4 * (x[i] - 1) ** 3 - 1)
    yinc = -s * 2 * (y[i] - 2)
    if xinc ** 2 + yinc**2 < d**2:
        break
    else:
        x[i+1] = x[i] + xinc; y[i + 1] = y[i] + yinc; k = k+1

plt.figure(figsize=(10,10))
plt.scatter(x,y)
```
